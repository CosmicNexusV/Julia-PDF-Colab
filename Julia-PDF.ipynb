# Build Official Julia PDF Documentation / 用于构建 Julia 官方 PDF 文档
Note: Must switch to Julia kernel / 注意：必须切换到 Julia 内核

Check version / 检查版本

```python
versioninfo()
```

Prepare environment / 准备环境

```python
# 1. Download source code (Note your VERSION, e.g., release-1.11) / 1. 下载源码 (注意对应你的 VERSION 版本，这里以 release-1.11 为例)
run(`git clone --depth 1 --branch release-1.11 https://github.com/JuliaLang/julia.git`)

# 2. Download Unicode data / 2. 下载数据表
mkpath("/content/julia/doc/doc")
run(`wget -qO /content/julia/doc/doc/UnicodeData.txt https://www.unicode.org/Public/15.0.0/ucd/UnicodeData.txt`)

# 3. Install necessary Linux LaTeX dependencies / 3. 安装必要的 Linux LaTeX 依赖库
run(`apt-get update -y`)
run(`apt-get install -y texlive-latex-base texlive-latex-extra texlive-luatex latexmk fonts-dejavu python3-pygments`)
```

Generate documentation / 生成文档

```python
# 4. Enter directory and start compiling! / 4. 进入目录并开始编译！
cd("julia/doc")
run(`julia --project=. make.jl pdf`)
```
